### 2. Import YOLO and Load a Model
We'll import the `YOLO` class and load a pre-trained model (e.g., `yolov8n.pt` for the nano model, which is fast and lightweight).

```python
from ultralytics import YOLO

# Load a pre-trained YOLOv8 nano model
model = YOLO('yolov8n.pt')
```

### 3. Run Inference
Now, let's run inference on a sample image. The `ultralytics` package provides a sample image by default, or you can pass a URL to an image.

```python
# Perform object detection on an image (using a sample URL)
results = model('https://ultralytics.com/images/bus.jpg')

# Display the results
for result in results:
    result.show()  # Display the image with bounding boxes
    # result.save(filename='result.jpg') # Uncomment to save to file
```

```python
# URL of a classic YOLO street scene test image (dog, bicycle, car)
new_image_url = 'https://raw.githubusercontent.com/pjreddie/darknet/master/data/dog.jpg'

# Run inference on the new image
new_results = model(new_image_url)

# Display the results
for result in new_results:
    result.show()
```
